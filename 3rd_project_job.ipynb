Step 3: Data Cleaning & Preprocessing
Key preprocessing steps include:




3.1 Handling missing values using mean, median, or mode

First, check the missing values:

```python
print(df.isnull().sum())
```

Numerical columns → Median

```python
numerical_columns = [
    'ssc_percentage',
    'hsc_percentage',
    'notice_period_days',
    'employment_gap_months'
]

for col in numerical_columns:
    df[col] = df[col].fillna(df[col].median())
```

```python
categorical_columns = [
    'career_switch_willingness',
    'relevant_experience',
    'job_role_match',
    'layoff_history',
    'relocation_willingness'
]

for col in categorical_columns:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
print(df.isnull().sum())
```

You can use mean instead:

Categorical columns → Mode
For example:

mean() → average value
median() → middle value
mode() → most frequently occurring value
fillna() → fills missing values

4.6 Correlation analysis among numeric features

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Calculate correlation
correlation = df.corr(numeric_only=True)

# Create figure with larger size
plt.figure(figsize=(16, 12))

# Create mask to hide upper triangle
mask = np.triu(np.ones_like(correlation, dtype=bool), k=1)

# Draw heatmap
sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    mask=mask,
    linewidths=0.5,
    square=True,
    annot_kws={"size": 9},
    cbar_kws={"shrink": 0.8},
    vmin=-1,
    vmax=1
)

plt.title("Correlation Analysis Among Numeric Features", fontsize=16, pad=20)

plt.xticks(rotation=45, ha="right", fontsize=10)
plt.yticks(rotation=0, fontsize=10)

plt.tight_layout()
plt.show()
```

5.1 Experience category (Fresher / Junior / Senior)




```python
import numpy as np

df['experience_category'] = np.select(
    [
        df['years_of_experience'] == 0,
        df['years_of_experience'].between(1, 3),
        df['years_of_experience'] >= 4
    ],
    [
        'Fresher',
        'Junior',
        'Senior'
    ],
    default='Unknown'
)

print(df[['years_of_experience',
          'experience_category']].head())
```

5.2 Academic Performance Bands

```python
df['academic_average'] = df[
    ['ssc_percentage',
     'hsc_percentage',
     'degree_percentage']
].mean(axis=1)
```

```python
df['academic_performance'] = pd.cut(
    df['academic_average'],
    bins=[-np.inf, 50, 75, np.inf],
    labels=['Low', 'Medium', 'High'],
    right=False
)

print(df[['academic_average',
          'academic_performance']].head())
```

5.3 Skills Match Level (Low / Medium / High)

```python
df['skills_match_level'] = pd.cut(
    df['skills_match_percentage'],
    bins=[-np.inf, 50, 75, np.inf],
    labels=['Low', 'Medium', 'High'],
    right=False
)

print(df[['skills_match_percentage',
          'skills_match_level']].head())
```

5.4 Interview Performance Category
